## POO: Programación Orientada a Objetos

En este paradigma se busca abstraer elementos del mundo real y tratar de conceptualizarlos en código.

### Conceptos Básicos de POO

- ***Clase:*** conjunto de objetos (plantilla)

Dentro de una clase podemos definir caracteristicas y comportamientos para todos los objetos del conjunto y la plantilla

- ***Objeto:*** Una instancia de la **Clase**
- ***Instancia:*** "Si la clase es el conjunto de Todos, el objeto ha de ser una unidad de ese todo", ha de referirse a la creación de un objeto a partir de una **clase** o **plantilla**

### Pilares de POO

- Abstraccion
- Encapsulamiento
- Herencia
- Polimorfismo

#### Concepto de Herencia

Hay una clase Base o "Súper" que ha de dar origen a subclases que conservarán sus características y métodos con la posibilidad de "agregar algo nuevo"



#### Concepto de Polimorfismo

- Indica que hay varias maneras en las que una clase o método se puede presentar, por ejemplo:
en **Animal** se encuentra **Gato** y **Perro**, pero a fin de cuentas **Gato** y **Perro** son de clase **Animal** en vista al código

En lo referente a los *métodos* se tienen dos casos especiales:

- *Sobre carga*: Múltiples métodos del mismo nombre pero con diferentes parámetros

        void sumar(int a, int b); // En vista al código a las dos funciones se les llamará de la misma manera
        void sumar(float a, float b);
- *Sustitución*: Un método de clase derivada reemplaza a la clase base



### Enfoques de resolucion de problemas

#### Enfoque Top-Down

Se va de lo general a lo específico. Primero pensamos en los procesos que se dan dentro de lo que queremos codificar, de tal manera que abstraigamos pasos y reglas básicas del negocio para luego hacer lo mismo para cada uno de dichos pasos hasta reducirlos lo suficiente para poder conceptualizarlos en pseudocódigo; Esto es algo parecido al "divide y vencerás". Este enfoque facilita un Diseño General estructural en nuestra aplicación además de permitirnos enfocarnos en lo básico del problema envés de complicarnos de más.

#### Enfoque Bottom Up

 Este enfoque es lo contrario, se empieza definiendo procesos o reglas básicas de negocio para luego entrelazarlas entre sí y formar una Idea compleja. Este enfoque permite definir y mejorar estos procesos de mejor manera y además facilita la innovación y extensibilidad del proyecto; aunque claro, cabe el riesgo de querer agregar procesos de más que se salen de los requerimientos originales de la aplicación

## Anotaciones clave antes de la clase

***Crear un Objeto:*** Hace referencia a la accion de crear una instancia de una clase, una clase ha de permitir crear múltiples objetos diferentes con mismas propiedades, pero con diferentes atributos; durante este proceso se debe destacar que se le ha de dar valor específico a las propiedades que tiene la clase. **Por ejemplo: Todo Vehículo tiene una marca y un tamaño, asi como todo átomo contiene un número atómico.**

***Atributo:*** El concepto de un atributo para un objeto ha de referirse al valor dado a una de las características propias (propiedad) de la clase plantilla para dicha instancia en específico.

Se debe recalcar que una clase contiene propiedades, mientras que es el objeto (una instancia de la clase) el que tiene un atributo (valor); para esta asignatura:

      atributo            ---> valor.
    instanciar una clase <---> crear un objeto

Solo ten en cuenta que todo **Objeto** debería de tener:

- Estado: obviamente ha de tener atributos.
- Comportamiento: guarda funciones o procedimientos.
- Identidad: debe pertenecer a una clase

***New:*** Palabra reservada necesaria para la creación de una clase, en los lenguajes de programación ha de encargarse de reservar espacios de un tipo de dato en memoria.

***POO en un programa:*** Se trata de un tipo de programa que contiene colecciones de objetos que colabora y cooperan entre sí para lograr cumplir un propósito final; es importante recalcar que las clases pueden relacionarse entre sí, inclusive desde el método de un objeto creado puedo crear una instancia de otra clase más.

***Abstraccion:*** Desmenuzar un tipo de objeto y conceptualizar solo lo más importante de él en código. Piensa ¿qué es lo que hace a un carro ser un carro? ¿Qué debería tener todo carro? Y no pongas nada más de lo necesario.

***Encapsulamiento:*** Debido a que en un programa las clases comunmente han de interactuar unas con otras y hasta intercambiar información, se debe optar por limitar el acceso hacia atributos sensibles por parte de una clase a otra o hacer en su defecto que una clase no pueda exponer información sensible donde no le corresponde. Para llevar a cabo todo esto se utilizan modificadores de acceso tanto en clases como en sus atributos:

    private: privado, no es accesible desde el exterior, sino solo desde adentro de la clase
    public: accesible desde afuero y desde adentro

## Idea de Encapsulamiento

Es uno de los pilares básicos de la Programación Orientada a Objetos. Se trata de proteger los datos y atributos de una clase, de tal forma que se pueda tener un control de quien accede a dicha información, especialmente si es delicada, mediante modificadores de acceso.

### Concepto Formal de Encapsulamiento

- Ocultamiento de datos: proteger atributos internos de un objeto.
- Agrupación: orientado a los datos y métodos de una clase.
- Control de acceso: evitar acceso directo.

## Atributos privados y métodos públicos

Cuando declaramos un atributo como privado dentro de una clase no podemos modificarlo o manejarlo directamente desde nuestra clase principal, sino mediante un método dentro de la clase que contiene dicho atributo.

Dentro de una clase cuyos atributos han sido encapsulados existiran métodos para obtener/ saber los valores de un atributo (get) y métodos encargados de modificar esos valores (set)

# Asociación y Multiplicidad

## Concepto de Asociación

Cuando hablabamos de la POO nos referiamos a un paradigma en el que las clases colaboran alrededor de una regla de negocio para generar un programa. Bueno, pues esto nos dice implicitamente que las clases tienden a interactuar entre sí, que se conocen pues, y que, por lo tanto, estas se asocian entre sí; ese es el concepto de Asociación de clases.

    Dos clases colaboran <----> Las dos clases estan asociadas

La Asociación tiene dos componentes muy importantes:

- Relación semántica: conexión lógica entre las clases (Una figura geometrica debe tener vértices por ejemplo)
- Relación Estructural: como se comunican y colaboran estas clases.

Con base en estos componentes un objeto puede estar asociado a otro o ambos objetos asociados entre sí:

    clase1 ---> clase2

Clase 2 conoce de Clase 1, más Clase 1 no conoce a Clase 2: Un inventario de guitarras tiene una guitarra, más la guitarra no tiene como propiedad algo relacionado con el inventario más que solo pertenecer a este.

    clase1 <---> clase2

Ambas clases se conocen mutuamente: Un paralelo tiene datos de los estudiantes matriculados dentro su aula, pero un estudiante también tiene datos de los paralelos a los cuales este pertenece.

## Concepto de Multiplicidad

La multiplicidad busca describir la relación de asociación que mantienen dos clases con base en una cuestión:

¿Cuántas instancias de una clase hay en la otra?

### UML

UML son las siglas de **"Unified Modeling Language"**, el cual es un lenguaje utilizado para modelar las relaciones que han de tener dos o más clases dentro del contexto del diseño de Software por medio de diagramas y para ello se tienen varias notaciones importantes, incluyendo:

#### La notación cardinal de puntos

    0----x

Indica directamente el número de instancias de una clase dentro de otra clase:

- 1---1: uno a uno (un carro con un motor)
- 1---n: uno a varios (un escritor puede ser autor de varios libros)
- n---m: muchos a muchos (Un estudiante con varios cursos y un curso con varios estudiantes)

# Clases concretas, abstractas e interfaces

## Anotaciones previas

****Clean Code**** :

- estandarización de variables: nombres correctos y descriptivos para variables, uso de formatos como *Camel Case*.
- Los métodos preferiblemente deben contar con el menor número posible de parámetros.

## Clase Concreta

Son las clases normales con las que hemos venido tratando, esta complemente implementada y tiene atributos, métodos y puede ser instanciada además de poder ser heredada o heredar de otras clases



## Clase Abstracta

Es una clase que no puede ser instanciada directamente, de tal manera que es utilizada de modelo para otras clases y puede contener métodos abstractos. Contiene atributos y métodos declarados; sin embargo, estos no estarán desarrollados hasta que otras clases los implementen.

Se complementa con la herencia y polimorfismo, ya que clases hijas surgidas de esta clase abstracta pueden implementar la definición de aquellos métodos abstractos de la clase por medio de un override

## Interfaz

Podemos asociar la interfaz a un contrato que debe cumplir toda clase que lo implemente, dentro de esta se define o se declara la existencia de métodos declarados, más no necesariamente implementados todavía.

Aquellas clases que implementan una interfaz se caracterizarán por contar con los mismos métodos que se declararon en una interfaz, quedan relacionadas de forma indirecta (herencia indirecta)

# Principios SOLID



Estas son cinco directrices importantes que se deben de seguir en el diseño de software para legar mejorar la mantenibilidad, flexibilidad y escalabilidad del software.



## Single Responsibility <Responsabilidad única>

Una clase debe encargarse de una responsabilidad única dentro del programa, dígase que una clase debe estar bien fundamentada y cohesionada en lo referente al porqué de su existencia, más no debe intentar hacer los procedimientos de otras clases. La idea es hacer una mejor de división del trabajo dentro de las clases del código

 Una clase que se adjudica más comportamientos de los necesarios lleva a un código más complicado de mantener y escalar, además que complica el proceso de análisis de errores para el programa.

Este principio busca en gran medida lograr una alta cohesión y bajo acoplamiento de las clases involucradas en el programa.



## Open/Close <Abierto/Cerrado>

Una clase debe poder ser ampliada, pero no modificada directamente en código cada vez que requiera escalar el programa con nuevas funcionalidades.

Si se requiere agregar nuevos comportamientos a una clase, se debe recurrir a una interfaz o una clase abstracta envés de agregar dicho método en la clase directamente. Recuerda que se busca construir un todo con las bases que ya fueron definidas.



El principio es muy aplicable en filtros de búsqueda, si se planea filtrar por varios parametros es recomendable crear una clase por cada parametro que hereden una forma de validar de una interfaz; por el momento parecerá inutil hacer estos casos por separado, pero con el siguiente principio nos daremos cuenta de que estos filtros al tener una misma firma podemos armar un sistema de búsqueda dinámica mediante una lista.

## Sustitución de Liskov

"*Todos para uno y uno para todos*"



Si se da el hecho de que sucedan casos de herencia en código, toda clase derivada debe poder ser reemplazada por su clase padre sin que se rompa la lógica del programa. Este principio está en gran medida relacionado con el polimorfismo <las diferentes formas o variantes que puede tener algo>.

Por ejemplo: De la clase *Ave* se heredan *Gallina* y *Pato*, a ojos del código tanto la clase *Gallina* como *Pato* deben poder ser consideradas también como clases de tipo *Ave*; dicho esto puedes hacer diferentes métodos que trabajen con un objeto de la clase *Ave* como parametro, lo que hace que un objeto de tipo *Gallina* o de tipo *Pato* también puedan ser aceptados envés de tener que definir un mismo método para las dos clases, todo esto se puede dar gracias que tanto *Gallina* como *Pato* son también un tipo de *Ave*.

Solo recuerda que: "*Toda gallina es un ave, pero no toda ave es una gallina*".

**Ojo**! Este principio también aplica para interfaces

## Interface Segregation <Segregcación de Interfaces>

Anteriormente, ya hemos visto que las interfaces pueden ampliar los comportamientos de ciertas clases. Este principio nos dice que apliquemos una simple responsabilidad para cada interfaz que vayamos a implementar, o sea, una sola interfaz no tiene por qué estar conteniendo cláusulas que no le conciernen debido a que si una clase la implementa también adquiriría una responsabilidad que no le concierne.



## Inversión de Dependencia

"*Un módulo de alto nivel no debe depender de un módulo de bajo nivel directamente, sino de su abstracción*"

Cuando se trata de un proyecto de programación grande, no se debería dar el caso de que una clase contenga a otra clase directamente.

    public class Cosa{
        /.. método importante ../
    }
    public class Wea {
        private Cosa _cosa = new Cosa();   // ❌ depende de la clase concreta
    }

Si no que lo que la clase debería de contener es una referencia abstracta o indirecta de ella, por medio de una interfaz:

    public interface ICosa{
        /.. firma de método importante ../
    }

    pulic class wea: ICosa{
        /.. def de metodo importante ../
    }
    public class Wea{
        private ICosa _cosa = new Cosa();    // wea polimorfica
    }

Lo que hemos hecho es basarnos del principio de Liskov y el polimorfismo de herencia: ahora la clase en cuestión almacena no solo una plantilla de objetos, sino un grupo de plantillas que implementan "*ICosa*", de tal manera que si la implementación de este "*método importante*" cambia algún día solo tenga que crear una nueva "*Cosa*" con esa nueva implementación mientras dejo intacta a "*Wea*".

***Nota:*** Este principio es utilizado ampliamente en la arquitectura de tipo Hexagonal, ya que está diseñada para adaptar y escalar un servicio de rápido crecimiento.

Comunmente se planea utilizar un método en concreto de la clase que se busca almacenar, con el caso quizas de un finder o un sender. Por lo que para implementar nuevas formas de realizar dicho procedimiento sin modificar el módulo de alto nivel este debe contener aunque sea una referencia indirecta a lo que planea utilizar, lo que puede ser una desventaja comparada con métodos de tipo estático. *Recomendación personal:* si sabes que es un proyecto pequeño: manten tus finders o senders en modo estático no sin antes heredar la firma una o varias firmas comunes de estos mediante interfaces si lo crees necesario <Abierto/Cerrado>.
