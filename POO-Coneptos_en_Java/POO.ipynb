## POO: Programación Orientada a Objetos

En este paradigma se busca abstraer elementos del mundo real y tratar de conceptualizarlos en código.

### Conceptos Básicos de POO

- ***Clase:*** conjunto de objetos (plantilla)

Dentro de una clase podemos definir caracteristicas y comportamientos para todos los objetos del conjunto y la plantilla

- ***Objeto:*** Una instancia de la **Clase**
- ***Instancia:*** "Si la clase es el conjunto de Todos, el objeto ha de ser una unidad de ese todo", ha de referirse a la creación de un objeto a partir de una **clase** o **plantilla**

### Pilares de POO

- Abstraccion
- Encapsulamiento
- Herencia
- Polimorfismo

#### Concepto de Herencia

Hay una clase Base o "Súper" que ha de dar origen a subclases que conservarán sus características y métodos con la posibilidad de "agregar algo nuevo"



#### Concepto de Polimorfismo

- Indica que hay varias maneras en las que una clase o método se puede presentar, por ejemplo:
en **Animal** se encuentra **Gato** y **Perro**, pero a fin de cuentas **Gato** y **Perro** son de clase **Animal** en vista al código

En lo referente a los *métodos* se tienen dos casos especiales:

- *Sobre carga*: Múltiples métodos del mismo nombre pero con diferentes parámetros

        void sumar(int a, int b); // En vista al código a las dos funciones se les llamará de la misma manera
        void sumar(float a, float b);
- *Sustitución*: Un método de clase derivada reemplaza a la clase base



### Enfoques de resolucion de problemas

#### Enfoque Top-Down

Se va de lo general a lo específico. Primero pensamos en los procesos que se dan dentro de lo que queremos codificar, de tal manera que abstraigamos pasos y reglas básicas del negocio para luego hacer lo mismo para cada uno de dichos pasos hasta reducirlos lo suficiente para poder conceptualizarlos en pseudocódigo; Esto es algo parecido al "divide y vencerás". Este enfoque facilita un Diseño General estructural en nuestra aplicación además de permitirnos enfocarnos en lo básico del problema envés de complicarnos de más.

#### Enfoque Bottom Up

 Este enfoque es lo contrario, se empieza definiendo procesos o reglas básicas de negocio para luego entrelazarlas entre sí y formar una Idea compleja. Este enfoque permite definir y mejorar estos procesos de mejor manera y además facilita la innovación y extensibilidad del proyecto; aunque claro, cabe el riesgo de querer agregar procesos de más que se salen de los requerimientos originales de la aplicación

## Anotaciones clave antes de la clase

***Crear un Objeto:*** Hace referencia a la accion de crear una instancia de una clase, una clase ha de permitir crear múltiples objetos diferentes con mismas propiedades, pero con diferentes atributos; durante este proceso se debe destacar que se le ha de dar valor específico a las propiedades que tiene la clase. **Por ejemplo: Todo Vehículo tiene una marca y un tamaño, asi como todo átomo contiene un número atómico.**

***Atributo:*** El concepto de un atributo para un objeto ha de referirse al valor dado a una de las características propias (propiedad) de la clase plantilla para dicha instancia en específico.

Se debe recalcar que una clase contiene propiedades, mientras que es el objeto (una instancia de la clase) el que tiene un atributo (valor); para esta asignatura:

      atributo            ---> valor.
    instanciar una clase <---> crear un objeto

Solo ten en cuenta que todo **Objeto** debería de tener:

- Estado: obviamente ha de tener atributos.
- Comportamiento: guarda funciones o procedimientos.
- Identidad: debe pertenecer a una clase

***New:*** Palabra reservada necesaria para la creación de una clase, en los lenguajes de programación ha de encargarse de reservar espacios de un tipo de dato en memoria.

***POO en un programa:*** Se trata de un tipo de programa que contiene colecciones de objetos que colabora y cooperan entre sí para lograr cumplir un propósito final; es importante recalcar que las clases pueden relacionarse entre sí, inclusive desde el método de un objeto creado puedo crear una instancia de otra clase más.

***Abstraccion:*** Desmenuzar un tipo de objeto y conceptualizar solo lo más importante de él en código. Piensa ¿qué es lo que hace a un carro ser un carro? ¿Qué debería tener todo carro? Y no pongas nada más de lo necesario.

***Encapsulamiento:*** Debido a que en un programa las clases comunmente han de interactuar unas con otras y hasta intercambiar información, se debe optar por limitar el acceso hacia atributos sensibles por parte de una clase a otra o hacer en su defecto que una clase no pueda exponer información sensible donde no le corresponde. Para llevar a cabo todo esto se utilizan modificadores de acceso tanto en clases como en sus atributos:

    private: privado, no es accesible desde el exterior, sino solo desde adentro de la clase
    public: accesible desde afuero y desde adentro

## Java: Clases y Objetos en código

Hemos visto anteriormente que las clases pueden interactuar entre sí, de tal manera que una clase dentro de sus métodos puede crear una instancia de otra clase. Como se veía en programación estructurada, siempre debía de haber un bloque principal que se encargue de coordinar los otros bloques, en el caso de la programación orientada a objetos debe haber siempre una clase principal con un método estático **main** que se encargue de coordinar las demás clases.

Antes de querer programar algo en Java talvéz quieres seguir los siguientes pasos:

- Crear un proyecto (para Java)
- Crear un paquete en la carpeta de SRC (los paquetes almacenan clases)
- Crear clases

Nota: La experiencia varía enormemente si utilizas un IDLE dedicado a **Java** como "Oracle" o "IntelliJ" a que si estuvieras usando "Visual Studio Code", en los primeros dos IDLE cada clase que se cree se tratará como un archivo diferente además de poder almacenarse en paquetes, mientras que con el último IDLE todas clases estarán en un mismo archivo de tipo **java**
